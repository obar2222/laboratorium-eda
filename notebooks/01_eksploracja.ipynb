# Eksploracja danych sprzedażowych

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "../data/sprzedaz.csv",
    sep=";"
)
df.head()
```

## Struktura danych

```python
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
print(df.duplicated().sum())
```

```python
df["wartosc"].describe()
```

## Niespójności w kolumnie `kanal`

```python
df["kanal"].value_counts()
```

Ten sam kanał zapisano na kilka sposobów (`Internet`, `internet`, `internet ` oraz `Sklep`, `sklep`, `SKLEP`) - różnią się wielkością liter i spacją na końcu.

## Ujednolicenie nazw kanałów

```python
df["kanal"] = (
    df["kanal"]
    .str.strip()
    .str.lower()
)
df["kanal"].value_counts()
```

## Podstawowe miary

```python
print("Liczba zamówień w kanałach:")
print(df.groupby("kanal").size())
print()
print("Suma wartości w kanałach:")
print(df.groupby("kanal")["wartosc"].sum())
print()
print("Średnia wartość zamówienia:", df["wartosc"].mean())
print("Mediana wartości zamówienia:", df["wartosc"].median())
```

## Wykres sprzedaży według kanału

```python
sprzedaz = df.groupby("kanal")[
    "wartosc"
].sum()

sprzedaz.plot(kind="bar")
plt.title("Suma sprzedaży według kanału")
plt.xlabel("Kanał")
plt.ylabel("Suma sprzedaży (zł)")
plt.xticks(rotation=0)
plt.show()
```

## Wnioski

1. Internet generuje 1490 zł, a sklep 760 zł - kanał internetowy odpowiada za ok. 66% sprzedaży, choć liczba zamówień jest taka sama (po 6).
2. Średnia wartość zamówienia w internecie (ok. 248,33 zł) jest prawie dwa razy wyższa niż w sklepie (ok. 126,67 zł); najwyższe zamówienia (310 zł i 400 zł) pochodzą z internetu.
